### Loading Data

First, let's load a sample dataset into a pandas DataFrame. If you have your data in a different format or location, please adjust the path and loading function accordingly.

```python
import pandas as pd

# Assuming your data is in a CSV file named 'data.csv'
# If your file is different, please update the path and file name.
try:
    df = pd.read_csv('data.csv')
    print("Data loaded successfully.")
    display(df.head())
except FileNotFoundError:
    print("Error: 'data.csv' not found. Please make sure the file is in the correct directory.")
    print("Creating a sample DataFrame for demonstration purposes.")
    data = {'Patient_ID': [1, 2, 3, 4, 5, 6],
            'Age': [30, 45, None, 60, 25, 50],
            'Gender': ['Male', 'Female', 'Male', 'Female', None, 'Male'],
            'Diagnosis': ['Flu', 'Cold', 'Fever', None, 'Flu', 'Cold'],
            'Treatment_Cost': [100, 150, 200, 120, None, 180]}
    df = pd.DataFrame(data)
    display(df.head())
```

### Identify Missing Values

# Q2. Handling Missing Age

Scenario:
 Age is critical for medical analysis, but some values are missing.

Task:

Replace missing Age values with an appropriate method

Justify your choice (mean/median

### Analyzing 'Age' Distribution for Imputation Strategy

Before imputing missing 'Age' values, it's crucial to understand the distribution of the existing 'Age' data. If the distribution is skewed (not symmetrical), the median is generally a better choice for imputation as it's less sensitive to outliers. If the distribution is roughly symmetrical, the mean can be used.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the distribution of the 'Age' column
fig = plt.figure(figsize=(8, 5))
sns.histplot(df['Age'].dropna(), kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

# Calculate mean and median for 'Age'
age_mean = df['Age'].mean()
age_median = df['Age'].median()

print(f"Mean Age: {age_mean:.2f}")
print(f"Median Age: {age_median:.2f}")
```

### Justification and Imputation of Missing 'Age' Values

Based on the distribution shown above, we can decide whether to use the mean or median. Since the sample data created earlier has a small number of entries and potentially some outliers, the **median** is a more robust choice for imputation as it is less affected by extreme values than the mean. The median represents the middle value of the data, which is often a good central tendency measure for skewed distributions or small datasets.

Let's fill the missing 'Age' values with the calculated median.

```python
# Replace missing 'Age' values with the median
df['Age'].fillna(df['Age'].median(), inplace=True)

print("Missing 'Age' values imputed with the median.")

# Verify that there are no more missing 'Age' values
print("Missing values after imputation:")
display(df.isnull().sum())
```

# Q3. Handling Missing Treatment Cost

Scenario:
 Treatment cost is highly skewed due to expensive treatments.

Task:

Handle missing Treatment_Cost values

Choose the correct imputation method and explain why

### Analyzing 'Treatment_Cost' Distribution

To appropriately handle missing `Treatment_Cost` values, we should first examine its distribution. Since the scenario states it's highly skewed, the median will be a more suitable imputation strategy than the mean, as the median is less affected by outliers and extreme values often found in skewed data.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize the distribution of the 'Treatment_Cost' column
fig = plt.figure(figsize=(8, 5))
sns.histplot(df['Treatment_Cost'].dropna(), kde=True)
plt.title('Distribution of Treatment Cost')
plt.xlabel('Treatment Cost')
plt.ylabel('Frequency')
plt.show()

# Calculate mean and median for 'Treatment_Cost'
treatment_cost_mean = df['Treatment_Cost'].mean()
treatment_cost_median = df['Treatment_Cost'].median()

print(f"Mean Treatment Cost: {treatment_cost_mean:.2f}")
print(f"Median Treatment Cost: {treatment_cost_median:.2f}")
```

### Justification and Imputation of Missing 'Treatment_Cost' Values

As observed from the histogram and confirmed by the scenario description, the `Treatment_Cost` distribution is likely skewed (especially if there are very expensive treatments). In such cases, the **median** is the preferred imputation method because:

1.  **Robustness to Outliers**: The median is less sensitive to extreme values or outliers compared to the mean. Since expensive treatments can significantly pull the mean upwards, the median provides a more representative 'typical' cost.
2.  **Preservation of Distribution**: Using the median helps preserve the shape of the skewed distribution better than the mean, preventing artificial shifts in the data's central tendency.

Therefore, we will replace the missing `Treatment_Cost` values with the calculated median.

```python
# Replace missing 'Treatment_Cost' values with the median
df['Treatment_Cost'].fillna(df['Treatment_Cost'].median(), inplace=True)

print("Missing 'Treatment_Cost' values imputed with the median.")

# Verify that there are no more missing 'Treatment_Cost' values
print("Missing values after imputation:")
display(df.isnull().sum())
```

Q5. Invalid Age Values (Data Quality Check)

Scenario:
 Some patients have unrealistic age values (e.g., >100 or <0).

Task:

Detect such records

Decide whether to remove or correct them


### Detecting Invalid Age Values

As per the scenario, we need to detect patient records where the 'Age' is unrealistic (e.g., less than 0 or greater than 100). Let's identify these records.

```python
# Detect invalid age records
invalid_age_records = df[(df['Age'] < 0) | (df['Age'] > 100)]

print(f"Number of records with invalid age values: {len(invalid_age_records)}")

if not invalid_age_records.empty:
    print("Sample of records with invalid age values:")
    display(invalid_age_records)
else:
    print("No records with invalid age values found.")
```

### Deciding on Handling Invalid Age Values

For invalid age values, the best approach depends on the dataset and the specific problem. Common strategies include:

1.  **Removing** these records: If the number of invalid records is small and they are clearly erroneous, removing them might be the simplest solution to maintain data quality.
2.  **Correcting** them: If ages are slightly off, or if there's a clear pattern (e.g., age entered as 120 but likely 20), we might correct them. However, for values like negative ages, imputation (e.g., using the mean or median of valid ages) might be more appropriate than a direct correction unless there's external information.

Given that 'Age' is a critical medical analysis factor and values like '<0' or '>100' are highly improbable for patient data in a standard context, I will **replace** these invalid values with `NaN` first and then **impute** them using the median of the valid age range. This ensures that we don't lose the entire record and also keep the age within a reasonable distribution.

Alternatively, if there are very few such records, they could be removed directly. For this demonstration, we'll impute them.

```python
import numpy as np

# Replace invalid age values with NaN
df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = np.nan

# Recalculate the median age based on *valid* ages for imputation
valid_age_median = df['Age'].median()

# Impute the NaN values (originally invalid ages) with the valid median
df['Age'].fillna(valid_age_median, inplace=True)

print("Invalid age values have been replaced with NaN and then imputed with the median of valid ages.")

# Verify the changes
print("Number of records with invalid age values after correction:")
invalid_age_records_after = df[(df['Age'] < 0) | (df['Age'] > 100)]
print(len(invalid_age_records_after))

print("DataFrame head after correcting invalid ages:")
display(df.head())
```

# Q6. Outlier Detection (Treatment Cost)

Scenario:
 Extreme treatment costs are affecting analysis.

Task:

Detect outliers using IQR method

Display number of outliers

### Detecting Outliers in 'Treatment_Cost' using IQR Method

The Interquartile Range (IQR) method is a robust way to identify outliers, especially in skewed distributions. It defines outliers as values that fall below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR, where Q1 is the first quartile, Q3 is the third quartile, and IQR = Q3 - Q1.

```python
# Calculate Q1 (25th percentile) and Q3 (75th percentile) for 'Treatment_Cost'
Q1 = df['Treatment_Cost'].quantile(0.25)
Q3 = df['Treatment_Cost'].quantile(0.75)

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define the upper and lower bounds for outliers
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): {Q1:.2f}")
print(f"Q3 (75th percentile): {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower Bound for Outliers: {lower_bound:.2f}")
print(f"Upper Bound for Outliers: {upper_bound:.2f}")
```

### Displaying Number of Outliers

Now, let's identify the records that fall outside these bounds and count how many outliers we have in the 'Treatment_Cost' column.

```python
# Identify outliers
outliers = df[(df['Treatment_Cost'] < lower_bound) | (df['Treatment_Cost'] > upper_bound)]

# Display the number of outliers
print(f"Number of outliers in 'Treatment_Cost': {len(outliers)}")

if not outliers.empty:
    print("Sample of 'Treatment_Cost' outliers:")
    display(outliers)
else:
    print("No outliers found in 'Treatment_Cost'.")
```

# Q7. Outlier Treatment

Scenario:
The business team wants to retain all records.

Task:

Apply capping (Winsorization) on Treatment_Cost

Use 5th and 95th percentile



### Applying Capping (Winsorization) to 'Treatment_Cost'

Since the business team wants to retain all records, we will use a technique called **Winsorization** (or capping). This method involves setting a specified percentile of the lowest values to a certain lower limit and a specified percentile of the highest values to an upper limit. For this task, we will use the 5th and 95th percentiles as the capping limits.

```python
# Calculate the 5th and 95th percentiles for 'Treatment_Cost'
lower_cap = df['Treatment_Cost'].quantile(0.05)
upper_cap = df['Treatment_Cost'].quantile(0.95)

print(f"5th percentile for Treatment_Cost: {lower_cap:.2f}")
print(f"95th percentile for Treatment_Cost: {upper_cap:.2f}")

# Apply capping (Winsorization)
df['Treatment_Cost'] = np.where(df['Treatment_Cost'] < lower_cap, lower_cap, df['Treatment_Cost'])
df['Treatment_Cost'] = np.where(df['Treatment_Cost'] > upper_cap, upper_cap, df['Treatment_Cost'])

print("\n'Treatment_Cost' column has been capped using the 5th and 95th percentiles.")
```
